```python
import numpy as np
def sin(x):
    return float(np.sin(x))
```

Imports numpy, then defined a function sin(), and then steals the numpy code for sine and returns the value returned by the numpy function.

```python
import numpy as np
def cos(x):
    return float(np.cos(x))
```

Imports numpy, then defined a function cos(), and then steals the numpy code for cosine and returns the value returned by the numpy function.

```python
def tabulate():
	for i in range(1, 1001):
		x = i*2*np.pi/1000
		print( x , sin(x) , cos(x))
```

```python
def firstTen():
    columns = ["x","sin(x)","cos(x)"]
    print(columns)
    for i in range(10):
        columns[0] = i+1
        columns[1] = sin(i+1)
        columns[2] = cos(i+1)
        print(columns)
```

```python
tabulate()
firstTen()
```
